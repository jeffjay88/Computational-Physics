# Computational Physics use case 1: [ N-body gravity ]

<img src="img/img1.png" width="50%" align="right">


Consider Newton’s law:
$$[ F = G \frac{m_1 m_2}{r^2} ]$$

For **two bodies**, this is solvable analytically.

For **three or more bodies**:
- No general analytical solution
- Numerical integration is required <br /> <br /> <br />

### One-statement explanation
Newton's law gives an exact force, but once more than two bodies interact, the equations become nonlinear, chaotic and coupled, making analytical solutions impossible and requiring numerical time-integration. <br /> <br />

### Computational thinking:
1. Write equations of motion
2. Discretize time
3. Step forward numerically
4. Track errors and conservation laws

# Terminal, shell, commands

<b>`Terminal:`</b> the window/interface where you interact with the computer using text instead of clicking icons.

#### Role:
- Accepts text input
- Displays text output
- Connects you to a <b>shell</b>

#### <font color="red">Think of it as the doorway to the command-line world</font>

<b>`Example:`</b> GNOME Terminal, macOS Terminal, Windows Terminal (WSL)

##### <font color="red">*The terminal itself does nothing intelligent; it just displays input/output.*</font>

---

<b>`Shell:`</b> a programming environment. It is the program running inside the terminal (a command-line interpreter).

#### Role:
- Interprets what you type
- Executes commands
- Manages processes and environment variables

#### Common UNIX shells:
- `sh` (Bourne shell)  
- `ksh` (Korn shell)  
- `csh` (C shell)  
- `bash` (Bourne Again Shell; common on Linux/macOS and Git Bash)

#### <font color="red">Think of it as the interpreter between you and the operating system kernel.</font>

---

<b>`Commands`</b> are instructions you give to the shell.<br />
They can be (i) Built-in shell commands, (ii) External programs or (iii) Scripts.

<img src="img/img5.png" width="100%" align="center"  style="margin-bottom: -80px; margin-top: 0px;">

`How they relate:` *You type a <b>command</b> into the <b>terminal</b> → the <b>shell</b> interprets it → the <b>kernel</b> executes it.* <br /> <br /> <br />

### Common beginner confusion
❌ “The terminal is the shell” → Not exactly<br />
✔ The shell runs inside the terminal

❌ “Linux commands” → Mostly shell commands


## <font color="red">TRY</font>
`ps -p $$`: This prints the shell process, not the terminal. <br />
`echo $SHELL`: This prints the shell binary path (eg. '/bin/bash')

### Unix internals: Kernel vs Shell

**Kernel (central core of the OS)**  
- Loaded into memory at boot time (may be dynamically linkable).  
- Manages **users, processes, and security**.  
- Controls **I/O access** to peripheral devices.  
- Mostly written in **C** to support portability.

**Shell (your command interpreter)**  
- The interface between the user and the kernel.  
- Runs as a UNIX program that **interprets commands** and **initiates user processes**.  
- Think of it as a *programming language* you use interactively in the terminal.

A common mental picture is a set of layers: **Applications → Shell → Kernel → Hardware**.

### UNIX filesystem essentials

**Key ideas**
- The filesystem is a **single tree** rooted at `/` (no drive letters like `C:`).  
- **Filenames are case sensitive** (`Data.txt` ≠ `data.txt`).  
- Physical devices (external disks, network shares) can be **mounted** anywhere in the tree.

**At the OS level, UNIX treats (almost) everything as a file**
- **Ordinary file**: typically text/binary data  
- **Directory**: a file that contains pointers to other files  
- **Special file**: a pointer to a device (e.g., disks, terminals)

**Productivity tip: TAB completion**
- Press `<TAB>` to auto-complete command names and paths.  
- Press `<TAB>` twice to show all possible completions.

## Absolute vs relative paths
An absolute path starts from <b>/</b> and is location-independent; a relative path starts from your <b>current directory</b> and depends on where you are.<br />

Relative paths allow:
- Code portability
- Running simulations on different machines
- Collaboration

### Example
A simulation script referencing:
```bash
data/initial_conditions.csv
```
works everywhere.

An absolute path:
```bash
/home/student/Desktop/data.csv
```
breaks immediately.

### File naming rules that save you headaches

Practical filename hygiene (especially important for reproducible science):

- Avoid embedded spaces (use `_` or `-` instead).  
- Prefer alphanumeric characters; be careful with shell metacharacters like `* ? > < / ; $ ! [ ] { } \` and quotes.  
- Be consistent with case (UNIX is case-sensitive).  
- Use leading `.` only for hidden files/folders.  
- Avoid starting names with `+` or `-` (can be interpreted as options).  
- Make names descriptive: `gravitywaves_data_1991_2020.nc` beats `data1.nc`.

### Files, permissions, and `chmod`

#### Reading permissions with `ls -l`
`ls -l` shows (among other things) the **permission bits** at the far left, e.g.

```text
-rw-rw-r--  1 user group  68 Jan 31 21:57 biglist.txt
```

- The **first character** indicates *file type*  
  - `-` ordinary file  
  - `d` directory  
  - (others exist: `b`, `c`, `l`, etc. for special cases)
- The next **9 characters** are permissions in three groups:
  - **owner** (user), **group**, **others**
  - each group has `r` (read), `w` (write), `x` (execute)

**Meaning of `rwx` depends on whether it’s a file or directory**
- For an ordinary file:  
  - `r` = read contents  
  - `w` = modify contents  
  - `x` = execute as a program/script
- For a directory:  
  - `r` = list directory contents  
  - `w` = create/remove files in the directory  
  - `x` = enter the directory (`cd`) / access files within

#### Octal notation (why `664`, `775`, etc. appear)
Each permission triple is a number:
- `r = 4`, `w = 2`, `x = 1`  
So:
- `rw- = 4 + 2 = 6`
- `r-x = 4 + 1 = 5`
- `rwx = 4 + 2 + 1 = 7`

Example: `chmod 775 myscript.sh` gives:
- owner: `7` (rwx)
- group: `7` (rwx)
- others: `5` (r-x)

#### `chmod` examples you should know
```bash
chmod 644 report.txt      # owner can write; everyone can read
chmod 755 run_model.sh    # executable script for everyone
chmod 775 .               # common for shared group folders (collab)
chmod o+w file.txt        # add write permission for others (symbolic)
chmod go-r data.nc        # remove read for group & others (symbolic)
chmod +x run.sh           # add execute for all (symbolic shorthand)
```

⚠️ **Safe habits** (especially for students)  
- Treat `rm` and `chmod` as *high-impact* commands.  
- Prefer: `ls` → double-check paths → then run destructive actions.  
- In research workflows, keep raw data read-only and write results to `outputs/` or `results/`.
